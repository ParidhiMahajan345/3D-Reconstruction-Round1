```python

import cv2
import os
import glob

# Define paths
VIDEO_DIR = os.path.join(BASE_DIR, "input_video")
FRAME_DIR = os.path.join(BASE_DIR, "frames")

# Find uploaded video
video_files = glob.glob(os.path.join(VIDEO_DIR, "*.mp4"))

if not video_files:
    raise FileNotFoundError("No MP4 video found in input_video folder!")

video_path = video_files[0]

# Open video
cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

duration = total_frames / fps

print("Video:", os.path.basename(video_path))
print("Resolution:", width, "x", height)
print("FPS:", fps)
print("Duration:", round(duration, 2), "seconds")

# Extract 1 frame per second
interval = 1
frame_interval = int(fps * interval)

count = 0
saved = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    if count % frame_interval == 0:
        filename = f"frame_{saved:04d}.jpg"
        filepath = os.path.join(FRAME_DIR, filename)

        cv2.imwrite(filepath, frame, [cv2.IMWRITE_JPEG_QUALITY, 95])
        saved += 1

    count += 1

cap.release()

print("\nFrame extraction completed!")
print("Total frames extracted:", saved)
print("Saved inside:", FRAME_DIR)
```

```python

import torch
from vggt.models.vggt import VGGT

print("Loading VGGT-1B model...")
print("Downloading checkpoint. This may take a few minutes.")

model = VGGT.from_pretrained("facebook/VGGT-1B")

model = model.eval().to("cuda")

print("\nVGGT MODEL LOADED SUCCESSFULLY!")
print("Device:", next(model.parameters()).device)
print("GPU:", torch.cuda.get_device_name(0))
```

```python

import os
import gc
import torch
import numpy as np

from vggt.utils.load_fn import load_and_preprocess_images

# Use every extracted frame
all_frames = sorted(frames)

print("Total frames selected:", len(all_frames))

assert len(all_frames) == 95, "Expected 95 frames!"

# Process in batches to manage GPU memory
BATCH_SIZE = 5

all_predictions = []

dtype = (
    torch.bfloat16
    if torch.cuda.get_device_capability()[0] >= 8
    else torch.float16
)

for start in range(0, len(all_frames), BATCH_SIZE):

    batch_paths = all_frames[start:start + BATCH_SIZE]

    print(f"\nProcessing frames {start + 1} to {start + len(batch_paths)}")

    images = load_and_preprocess_images(batch_paths).to("cuda")

    with torch.inference_mode():
        with torch.autocast("cuda", dtype=dtype):
            predictions = model(images)

    batch_result = {
        key: value.detach().cpu()
        for key, value in predictions.items()
        if torch.is_tensor(value)
    }

    all_predictions.append(batch_result)

    del images, predictions, batch_result
    gc.collect()
    torch.cuda.empty_cache()

print("\nALL 95 FRAMES PROCESSED!")
```

```python

import torch
import gc
import os

from vggt.utils.load_fn import load_and_preprocess_images

# Release previous batch outputs
if "all_predictions" in globals():
    del all_predictions

gc.collect()
torch.cuda.empty_cache()

# Load ALL 95 frames in their original order
all_frames = sorted(frames)

print("Loading all", len(all_frames), "frames...")

images = load_and_preprocess_images(all_frames).to("cuda")

print("Input shape:", images.shape)

# Run one unified multi-view inference
dtype = torch.float16

print("\nStarting unified VGGT reconstruction...")

try:
    with torch.inference_mode():
        with torch.autocast("cuda", dtype=dtype):
            full_predictions = model(images)

    print("\nUNIFIED RECONSTRUCTION SUCCESSFUL!")

    # Save complete predictions
    save_path = os.path.join(
        OUTPUT_DIR, "vggt_95_frames_full_scene.pt"
    )

    cpu_predictions = {
        key: value.detach().cpu()
        for key, value in full_predictions.items()
        if torch.is_tensor(value)
    }

    torch.save(cpu_predictions, save_path)

    print("\nPrediction outputs:")
    for key, value in cpu_predictions.items():
        print(key, tuple(value.shape))

    print("\nSaved:", save_path)

except torch.cuda.OutOfMemoryError:
    print("\nGPU MEMORY LIMIT REACHED.")
    print("No unified reconstruction was saved.")
    print("We will adjust the inference strategy.")

finally:
    if "images" in globals():
        del images

    gc.collect()
    torch.cuda.empty_cache()
```

```python

import torch
import numpy as np
import os
import gc

# Load saved predictions
file_path = "/content/drive/MyDrive/BharatXR_3D_Reconstruction/outputs/vggt_95_frames_full_scene.pt"
output_path = "/content/drive/MyDrive/BharatXR_3D_Reconstruction/outputs/BharatXR_95_Frames_PointCloud.ply"

data = torch.load(file_path, map_location="cpu", weights_only=True)

# Extract 3D points, confidence and images
points = data["world_points"][0].numpy()
confidence = data["world_points_conf"][0].numpy()
images = data["images"][0].numpy()

# Convert images from [N, 3, H, W] to [N, H, W, 3]
images = np.transpose(images, (0, 2, 3, 1))

# Prepare RGB colours
if images.min() < 0:
    images = (images + 1.0) / 2.0

images = np.clip(images, 0, 1)
colors = (images * 255).astype(np.uint8)

# Reduce point count while retaining all 95 frames
stride = 3

points = points[:, ::stride, ::stride, :]
confidence = confidence[:, ::stride, ::stride]
colors = colors[:, ::stride, ::stride, :]

# Use the 60th percentile as an initial confidence filter
threshold = np.percentile(confidence, 60)

mask = (
    (confidence >= threshold) &
    np.isfinite(points).all(axis=-1)
)

# Collect filtered points and colours
xyz = points[mask].astype(np.float32)
rgb = colors[mask].astype(np.uint8)

print("Confidence threshold:", threshold)
print("Total exported points:", len(xyz))

# Create binary PLY data
vertex = np.empty(
    len(xyz),
    dtype=[
        ("x", "<f4"),
        ("y", "<f4"),
        ("z", "<f4"),
        ("red", "u1"),
        ("green", "u1"),
        ("blue", "u1")
    ]
)

vertex["x"] = xyz[:, 0]
vertex["y"] = xyz[:, 1]
vertex["z"] = xyz[:, 2]
vertex["red"] = rgb[:, 0]
vertex["green"] = rgb[:, 1]
vertex["blue"] = rgb[:, 2]

# Write PLY file
header = f"""ply
format binary_little_endian 1.0
element vertex {len(vertex)}
property float x
property float y
property float z
property uchar red
property uchar green
property uchar blue
end_header
"""

with open(output_path, "wb") as f:
    f.write(header.encode("ascii"))
    vertex.tofile(f)

print("\nSUCCESS! 3D POINT CLOUD EXPORTED")
print("Saved at:", output_path)
print("File size:", round(os.path.getsize(output_path) / (1024 * 1024), 2), "MB")
```
